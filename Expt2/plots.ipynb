```julia
using Plots, CSV, DataFrames, LaTeXStrings, LsqFit
```

```julia
A = CSV.read("intercept.csv", DataFrame)
B = CSV.read("loop_width.csv", DataFrame)
C = CSV.read("tip-tip.csv", DataFrame)
```

```julia
lin(x,p) = @. p[1]*x + p[2]
```

```julia
scatter(C[!,1], C[!,2])
scatter!(A[!,1], A[!,2], ylim=(2,6))
scatter(B[!,1], B[!,2], ylim=(0,5), c="black", marker=:uptriangle, label="Data Point")

fiit = curve_fit(lin, B[!,1][end-3:end],  B[!,2][end-3:end], [0.3,2.4])
plot!(50:350, lin(50:350, fiit.param), linestyle=:dash, c="black", label="Linear Fit", ylim=(1.5,4.0))
plot!(xlabel=L"H_a \ (\mathrm{gauss})", ylabel=L"\mathrm{2 \times \ Loop \ Width}\ (\mathrm{V})")
```
