```julia
using Plots, LaTeXStrings, LsqFit
```

```julia
I = [0, 0.15, 0.33, 0.44, 0.62, 0.79, 1, 1.23, 1.5, 1.74, 2.09, 2.65, 2.91, 3.29, 3.51, 3.74]
B = [-224, -481, -708, -888, -1203, -1526, -1933, -2350, -2860, -3270, -3860, -4870, -5190, -5720, -6030, -6360]
linfit(x,p)= @. p[1]*x + p[2]
```

```julia
fiit = curve_fit(linfit, I, -B, [2.3,0.4]) 
fiit.param
```

```julia
standard_errors(fiit)
```

```julia
plot(I, linfit(I, fiit.param), markerstrokewidth=2, c="black", label=L"linear fit slope = $1681.53 \pm 20.08$")
scatter!(I, -B, marker=:rtriangle,ms=7, linestyle=:dash, xlabel= L"Current $\mathrm{I}$", ylabel=L"Magnetic Field $\mathrm{B\ \  (gauss)}$", label="data" , markercolor=RGBA(1, 1, 1, 0), markerstrokewidth=2, xticks=0:0.5:3.5, left_margin=4Plots.mm)
```
